# 02 - Exploratory Data Analysis

Explores salary patterns, job-market composition, and skill demand using the cleaned dataset produced in notebook 01.

```python
import sys, os
sys.path.append(os.path.abspath('../src'))
import pandas as pd

df = pd.read_csv('../data/processed/jobs_clean.csv')
df['Skills List'] = df['Skills List'].apply(eval)
df.shape
```

## Salary Overview

```python
df['Salary_USD'].describe()
```

```python
import plotly.express as px
px.histogram(df, x='Salary_USD', nbins=50, title='Salary Distribution (USD/year)')
```

## Salary by Experience, Role, Location, Industry, Company Size

```python
df.groupby('Experience Level')['Salary_USD'].mean().reindex(['Entry','Mid','Senior','Lead'])
```

```python
df.groupby('Job Title')['Salary_USD'].mean().sort_values(ascending=False)
```

```python
df.groupby('Location')['Salary_USD'].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby('Industry')['Salary_USD'].mean().sort_values(ascending=False)
```

```python
df.groupby('Company Size')['Salary_USD'].mean().sort_values(ascending=False)
```
